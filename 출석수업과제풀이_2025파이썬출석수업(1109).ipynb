```python
# 계산 목록(숫자1, 연산자, 숫자2)
calculations = [
    (10, '+', 5),
    (20, '-', 8),
    (6, '*', 7),
    (15, '/', 3),
    (8, '/', 0),   # 0으로 나누기
    (12, '%', 5),  # 잘못된 연산자
]
print("계산 결과: ")

# 여기에 들어갈 코드를 작성하세요
for i in range(len(calculations)):
    n1 = calculations[i][0]
    op = calculations[i][1]
    n2 = calculations[i][2]

    if op == '+':
        res = n1 + n2
        print(n1,op, n2, "=", res)
    elif op == '-':
        res = n1 - n2
        print(n1, op, n2, "=", res)
    elif op == '*':
        res = n1 *n2
        print(n1, op, n2, "=", res)
    elif op == '/':
        if n2 == 0:
            print("0으로 나눌 수 없습니다")
        else:
            res = n1 / n2
            print(n1, op, n2, "=", res)
    else:
        print("잘못된 연산자입니다")
```
